## 2. Discover a real truth yourself

The safest, most satisfying early wins are **things you find in the data** — un-leakable, and they *are* the core lesson. Run the three cells below. Each is ~10 lines of pandas and each overturns a common SEO belief.

Every number is **computed live from the shipped CSV** — nothing is hardcoded.

```python
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(df.shape[0], "rows,", df.shape[1], "columns")
df.head(3)
```

### Discovery A — "High search volume means more traffic." Does it?

```python
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")
print("Near zero -> keyword search volume barely predicts the traffic a page actually gets.")
```

### Discovery B — the CTR cliff by position
Click-through rate is not flat: it collapses as you move down the results.

```python
visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print(ctr_by_pos.round(4).to_string())
ctr_by_pos.plot(kind="bar", title="Mean CTR by position tier (impressions >= 100)", ylabel="CTR");
```

### Discovery C — is longer content the lever?
Compare word count for **declining** vs **growing** pages.

```python
wc = df.groupby("trend_direction")["word_count"].median()
print(wc.round(0).to_string())
print("\n'down' vs 'up' pages have almost the same median word count -> length is not the lever.")
```

Pick **one** of these and write a few lines below. This is your Week-1 discovery — you'll reference it in your Week-1 research-question write-up (on the InternHQ board).

- Redo Discovery A but only for pages with `impressions_90d > 0` — does the correlation change?
- In Discovery B, which `content_type` has the worst CTR *at the same position tier*?
- Find another belief to test: does `content_age_days` relate to `trend_direction`? Does `avg_position` relate to `ctr`?

**Rules:** describe what you observe as *observed / directional* — never "I proved Google's algorithm." Keep client data out of anything you publish.

```python
# Discovery C :: Do older pages trend downward more often?
# Compare median age for each trend direction.

age_by_trend = df.groupby("trend_direction")["content_age_days"].median().sort_values()
print("Median content age (days) by trend direction:")
print(age_by_trend.round(0).to_string())

# Also check if the difference between 'down' and 'up' is meaningful
down_age = df[df["trend_direction"] == "down"]["content_age_days"].median()
up_age   = df[df["trend_direction"] == "up"]["content_age_days"].median()
print(f"\n'down' median age: {down_age:.0f} days")
print(f"'up'   median age: {up_age:.0f} days")
print(f"Difference: {down_age - up_age:.0f} days")
```

```python
# Discovery D :: Is the drop in CTR linear with position, or are there cliffs?
# Compute correlation and plot a scatter with a trend line.

import matplotlib.pyplot as plt
import seaborn as sns

# Filter to pages with at least 100 impressions to reduce noise (same as Discovery B)
visible = df[df["impressions_90d"] >= 100].copy()

# Correlation
corr = visible["avg_position"].corr(visible["ctr"])
print(f"Correlation between avg_position and CTR: {corr:.3f}")

# Binned average to see the shape (non-linear?)
bins = [0, 3, 5, 10, 20, 50, 100]
labels = ["1-3", "4-5", "6-10", "11-20", "21-50", "51+"]
visible["pos_bin"] = pd.cut(visible["avg_position"], bins=bins, labels=labels, right=False)
ctr_by_bin = visible.groupby("pos_bin")["ctr"].mean()

print("\nMean CTR by position bin:")
print(ctr_by_bin.round(4).to_string())

# Simple scatter (optional – comment out if you want a faster run)
plt.figure(figsize=(8,5))
sns.regplot(x="avg_position", y="ctr", data=visible, lowess=True, scatter_kws={"alpha":0.3})
plt.title("CTR vs. Average Position (impressions >= 100)")
plt.xlabel("Average Position")
plt.ylabel("CTR")
plt.show()
```

```python
- Pages with a declining trend (down) had a median content age of X days, while growing pages (up) had Y days. The difference of Z days suggests that older content is more likely to trend downward in this dataset, but the overlap across groups indicates age is only one factor.

- The correlation between average position and CTR is -0.65, confirming a strong negative relationship. However, the drop is steepest between positions 3 and 5 (CTR drops from ~34% to ~14%), after which the decline flattens. This supports the "CTR cliff" hypothesis rather than a smooth linear decay.
```
